# Práctica #9: Distribución de Datos con plt.subplots()
**Objetivo**

Utilizar plt.subplots() para visualizar diferentes distribuciones generadas con NumPy y almacenadas en un DataFrame.

Instrucciones:
* Genera un DataFrame con Pandas que contenga 1000 valores en cuatro columnas (Normal, Uniforme, Exponencial, Poisson).
* Crea una figura con plt.subplots(2, 2) para organizar los histogramas en una cuadrícula de 2x2.
* Grafica cada distribución en un subplot usando ax[i, j].hist().
* Agrega títulos a cada subplot y ajusta la disposición con tight_layout().

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

#Generar los valores aleatorios para cada distribución
valnor = np.random.normal(0, 1, 1000)
valuni = np.random.uniform(-3, 3, 1000)
valexp = np.random.exponential(1, 1000)
valpoi = np.random.poisson(3, 1000)

#Crear la figura 2x2 y definir el tamaño
fig, ax = plt.subplots(2, 2, figsize=(8, 6))

#Crear el dataframe con los valores de cada distribución
df = pd.DataFrame({
    'Distribucion normal': valnor,
    'Distribucion uniforme': valuni,
    'Distribucion exponencial': valexp,
    'Distribucion poisson': valpoi
}
)

#Graficar los histogramas para cada distribución, bins representa el numero de barras, ax[i,j] ubicación de las graficas
ax[0, 0].hist(df['Distribucion normal'], bins=30, color="b", alpha=0.3)
ax[0, 0].set_title("Distribución Normal")

ax[0, 1].hist(df['Distribucion uniforme'], bins=30, color="c", alpha=0.5)
ax[0, 1].set_title("Distribución Uniforme")

ax[1, 0].hist(df['Distribucion exponencial'], bins=30, color="m", alpha=0.7)
ax[1, 0].set_title("Distribución Exponencial")

ax[1, 1].hist(df['Distribucion poisson'], bins=30, color="y", alpha=0.9)
ax[1, 1].set_title("Distribución Poisson")

#plt.tight_layout ajusta el espacio entre las graficas para que no se traslapen
plt.tight_layout()
#Mostrar las graficas
plt.show()
```
