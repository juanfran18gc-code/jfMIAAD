# Práctica #10: Análisis de Ventas con GridSpec
**Objetivo**

Utilizar GridSpec para organizar subplots de diferentes tamaños en un análisis de ventas.

Instrucciones:
1.  Genera un DataFrame de ventas con Mes, Ventas, y Gastos. Utiliza randint para generar números aleatorios enteros. Para ventas los valores deben estar entre 2000 y 5000. Para gastos, los valores debn estar entre 1000 y 4000.
2.  Usa GridSpec para organizar los subplots de la siguiente manera:
    * Primer gráfico (barra, ocupando dos columnas)
    * Segundo gráfico (línea, una columna)
    * Tercer gráfico (dispersión, dos filas).
3.  Agrega títulos y ajusta espacios.


```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import pandas as pd


meses = [
    "enero", "febrero", "marzo", "abril", "mayo", "junio",
    "julio", "agosto", "septiembre", "octubre", "noviembre", "diciembre"
]
#Generar los valores aleatorios para las ventas y gastos
ventas = np.random.randint(2000,5000, 12)
#print(ventas)
gastos = np.random.randint(1000,4000, 12)
#print(gastos)

#Crear la figura y definir el tamaño
fig = plt.figure(figsize=(16, 8))
# Definir una cuadrícula de 3x2
gs = gridspec.GridSpec(3, 2)

ax1 = fig.add_subplot(gs[0, :])  # Ocupa dos columnas
ax2 = fig.add_subplot(gs[1, 0])   # Ocupa una columna
ax3 = fig.add_subplot(gs[1:, 1])  # Ocupa dos filas


#Crear el dataframe con los valores de cada distribución
df = pd.DataFrame({
    'Mes': meses,
    'Ventas': ventas,
    'Gastos': gastos
}
)

#Grafica de barras Ventas y gastos por mes
ax1.bar(df['Mes'],df['Ventas'], color="b", alpha=0.3)
ax1.bar(df['Mes'],df['Gastos'], color="c", alpha=0.5)
ax1.set_title("Ventas y gastos por mes")

#Grafica de línea tendencia de ventas y gastos
ax2.plot(df['Mes'],df['Ventas'], color="r", marker="^" )
ax2.plot(df['Mes'],df['Gastos'], color="m", marker="s" )
ax2.set_title("Tendencia de Ventas y gastos por mes")
ax2.tick_params(axis='x', rotation=90)

#Grafica de dispersión de gastos y ventas
ax3.scatter(df['Ventas'],df['Gastos'], color="y", alpha=0.7)
ax3.set_title("Ventas vs Gastos")
ax3.set_xlabel("Ventas")
ax3.set_ylabel("Gastos")
#ax3.tick_params(axis='x', rotation=90)

#plt.tight_layout ajusta el espacio entre las graficas para que no se traslapen
plt.tight_layout()
#Mostrar las graficas
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import pandas as pd


meses = [
    "enero", "febrero", "marzo", "abril", "mayo", "junio",
    "julio", "agosto", "septiembre", "octubre", "noviembre", "diciembre"
]
#Generar los valores aleatorios para las ventas y gastos
ventas = np.random.randint(2000,5000, 12)
#print(ventas)
gastos = np.random.randint(1000,4000, 12)
#print(gastos)

#Crear la figura y definir el tamaño
fig = plt.figure(figsize=(16, 8))
# Definir una cuadrícula de 2X3 ajustando el ancho y alto relativo
gs = gridspec.GridSpec(2, 3, width_ratios=[2, 1, 1], height_ratios=[1, 2])

ax1 = fig.add_subplot(gs[0, :2])  # Ocupa dos columnas
ax2 = fig.add_subplot(gs[0, 2])   # Ocupa una columna
ax3 = fig.add_subplot(gs[1, :])  # Ocupa dos filas


#Crear el dataframe con los valores de cada distribución
df = pd.DataFrame({
    'Mes': meses,
    'Ventas': ventas,
    'Gastos': gastos
}
)

fig.suptitle("Análisis de ventas", fontsize=16)  # Título global

#Grafica de barras Ventas y gastos por mes
ax1.bar(df['Mes'],df['Ventas'], color="b", alpha=0.3)
ax1.bar(df['Mes'],df['Gastos'], color="c", alpha=0.5)
ax1.set_title("Ventas y gastos por mes")

#Grafica de línea tendencia de ventas y gastos
ax2.plot(df['Mes'],df['Ventas'], color="r", marker="^" )
ax2.plot(df['Mes'],df['Gastos'], color="m", marker="s" )
ax2.set_title("Tendencia de Ventas y gastos por mes")
ax2.tick_params(axis='x', rotation=90)

#Grafica de dispersión de gastos y ventas
ax3.scatter(df['Ventas'],df['Gastos'], color="y", alpha=0.7)
ax3.set_title("Ventas vs Gastos")
ax3.set_xlabel("Ventas")
ax3.set_ylabel("Gastos")
#ax3.tick_params(axis='x', rotation=90)

#plt.tight_layout ajusta el espacio entre las graficas para que no se traslapen
plt.tight_layout()
#Mostrar las graficas
plt.show()
```
