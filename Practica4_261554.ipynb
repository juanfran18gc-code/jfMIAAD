# Práctica #4 - Unidad 1. Manipulación de matrices
Crea dos matrices de 3x3 con números aleatorios entre 1 y 10. Realiza las siguientes operaciones:

1.  Suma ambas matrices.
2.  Resta la segunda matriz de la primera.
3.  Multiplica la primera matriz por 2.
4.  Realiza una multiplicación matricial entre las dos matrices.



```python
import numpy as np

print("Matrices de 3X3 con numeros aleatorios entre 1 y 10")
#Agregamos una semilla 2 para generar los mismos numeros aleatorios de A - reproducibilidad
np.random.seed(2)

#Generamos dos matrices de números aleatorios entre 1 y 10
A = np.random.randint(1,11,(3,3))

#Agregamos una semilla 3 para generar los mismos numeros aleatorios de B - reproducibilidad
np.random.seed(3)
B = np.random.randint(1,11,(3,3))
#Imprimimos las matrices generadas
print(A)
print(B)
#Obtenemos la suma de matrices
print("Suma de A + B: ", A + B)

#Obtenemos la resta de matrices
print("Resta de B - A: ", B - A)

#Obtenemos el producto de A * 2
print("Producto de A * 2: ", A * 2)

#Obtenemos la multiplicación matricial con np.dot
print("Multiplicación matricial de A y B: ", np.dot(A,B))
```
