# Práctica #3- Unidad 1. Creación de un arreglo y operaciones básicas

Crea un arreglo unidimensional de 20 elementos con valores aleatorios entre 1 y 100. Luego, realiza las siguientes operaciones:

1.   Encuentra el valor máximo del arreglo.
2.   Encuentra el valor mínimo del arreglo.
3.   Calcula la media del arreglo.
4.   Suma 10 a cada elemento del arreglo.





```python
import numpy as np

print("Arreglo de 20 elementos con valores enteros aleatorios entre 1 y 100")
#Agregamos una semilla 2 para generar los mismos numeros aleatorios= reproducibilidad
np.random.seed(2)

#Generamos un arreglo de números aleatorios entre 1 y 100
arreglo = np.random.randint(1,101,size=20)
#Imprimimos los resultados
print(arreglo)
#Obtenemos el valor máximo del arreglo
print("Valor máximo: ", arreglo.max())

#Obtenemos el valor mínimo del arreglo
print("Valor mínimo: ", arreglo.min())

#Obtenemos el promedio del arreglo
print("Media: ", arreglo.mean())

#Arreglo + 10 suma a cada elemento el valor
print("Sumar 10 a los elementos: ", arreglo + 10)
```
