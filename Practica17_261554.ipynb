```python
# Importación de bibliotecas
import numpy as np
import pandas as pd
#Cargamos la base de datos
df = pd.read_csv('/content/drive/MyDrive/Unidad02/nls97.csv')

#Imprimir el tamaño del dataset
print("Tamaño del conjunto de datos: cantidad de filas y columnas")
print(df.shape)
print("\n")
print("Información del conjunto de datos")
print(df.info())
```

# Ejercicio 01: Errores Tipográficos y Valores Inconsistentes: maritalstatus

1.  Imprime de la columna 'maritalstatus' los valores únicos.
2.  Imprime la frecuencia de cada valor único
3.  Cambia el valor 'Never-married' a 'Single'
4.  Imprime la frecuencia de cada valor único.

```python
#Imprimir valores únicos de columna
print(f'Valores únicos:', df['maritalstatus'].unique())

#Imprimir frecuencia de valores únicos
print(f'Frecuencia de Valores únicos:\n',df['maritalstatus'].value_counts())

#Reemplazar valor 'Never-married' a 'Single'
df['maritalstatus'] = df['maritalstatus'].replace('Never-married','Single', regex=False)

#Imprimir frecuencia de valores únicos
print(f'Frecuencia de Valores únicos:\n',df['maritalstatus'].value_counts())
```

# Ejercicio 02: Errores Tipográficos y Valores Inconsistentes: highestdegree.

1.  Imprime de la columna 'highestdegree' los valores únicos.
2.  Imprime la frecuencia de cada valor único
3.  Observa que cada valor categórico empieza con 0. Modifica las categorías para que todas estén en minúsculas y ninguna empiece con un número. Por ejemplo, si dice "2. High School", debería ser 'high school.
4.  Los cambios se realizarán en la misma columna 'highestdegrees'.
5.  Imprime los nuevos valores únicos.

```python
#Imprimir valores únicos de columna
print(f'Valores únicos:', df['highestdegree'].unique())

#Imprimir frecuencia de valores únicos
print(f'Frecuencia de Valores únicos:\n',df['highestdegree'].value_counts())

#Reemplazar categorías a minusculas y sin números iniciales
df['highestdegree'] = df['highestdegree'].str.lower().replace({r'^\d\.':''}, regex=True).str.strip()

#Imprimir valores únicos de columna modificada
print(f'Valores únicos modificados:', df['highestdegree'].unique())
```

# Ejercicio 03: Alta Cardinalidad - Frecuencias Bajas

*   De la columna 'maritalstatus' une las categorías Widowed y 'Single' en una categoría más general llamada 'Single/No Partner'. Crea una nueva columna, 'marital_grouped' donde aparezcan las nuevas categorías de maritalstatus.
*   Imprime los valores únicos de la nueva columna 'marital_grouped'.
*   Imprime la frecuencia de los valores únicos de la nueva columna 'marital_grouped'

```python
#Crear columna agrupada y Reemplazar las categorías Widowed y Single por 'Single/No Partner'
df['marital_grouped'] = df['maritalstatus'].replace({'Widowed':'Single/No Partner','Single':'Single/No Partner'})

#Imprimir valores únicos de columna agrupada
print(f'Valores únicos marital_grouped:', df['marital_grouped'].unique())

#Imprimir frecuencia de valores únicos de columna agrupada
print(f'Frecuencia de Valores únicos marital_grouped:\n',df['marital_grouped'].value_counts())
```

# Ejercicio 04: Tipos de Datos

*   Imprime la cantidad de bytes de dicha columna

*   En caso de haber realizado la modificación del tipo de datos, imprime la cantidad de bytes que ocupa dicha columna con el nuevo tipo de datos


```python
#Imprimir el tipo de dato de gender
print(f'Tipo de dato de gender:',{df['gender'].dtype})

#Imprimir cantidad de bytes de gender
print(f'\nMemoria total en bytes de gender:{df['gender'].memory_usage(deep=True)}')

#Modificar tipo de dato a category
df['gender'] = df['gender'].astype('category')

#Imprimir cantidad de bytes de gender como category
print(f'\nMemoria total en bytes de gender como tipo category:{df['gender'].memory_usage(deep=True)}')
```
