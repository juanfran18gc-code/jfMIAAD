###Conversión a minúsculas

```python
def convertir_minusculas(texto: str) -> str:
    """
    Convierte texto a minúsculas.

    Args:
        texto: Texto a convertir

    Returns:
        Texto en minúsculas
    """
    if pd.isna(texto):
        return ""
    return str(texto).lower()
```

###Eliminación de puntuaciones

```python
def eliminar_puntuacion(texto: str) -> str:
    """
    Elimina los signos de puntuación del texto.

    Args:
        texto: Texto a procesar

    Returns:
        Texto sin puntuación
    """
    if pd.isna(texto):
        return ""
    clean_text = re.sub(r'[^\w\s]', '', texto)
    return str(clean_text)
```

###Eliminación de emoticons/emojis


```python
def eliminar_emojis(texto: str) -> str:
    """
    Elimina completamente emoticons y emojis del texto usando la librería emoji.

    Args:
        texto: Texto con emoticons/emojis

    Returns:
        Texto sin emoticons/emojis

    """
    if pd.isna(texto):
        return ""

    resultado = str(texto)
    # La función replace_emoji reemplaza todos los emojis con el string especificado
    resultado = emoji.replace_emoji(resultado, replace='')

    return resultado.strip()
```

###Conversión de emoticons/emojis en palabras


```python
def convertir_emojis_a_palabras(texto: str, idioma: str = 'es') -> str:
    """
    Convierte emojis a palabras descriptivas usando la librería emoji.

    Args:
        texto: Texto con emojis
        idioma: Idioma para la conversión ('es', 'en', 'pt', 'it', 'fr', 'de')

    Returns:
        Texto con emojis convertidos a palabras

    """
    if pd.isna(texto):
        return ""

    resultado = str(texto)

    # Usar la función demojize de la librería emoji
    resultado = emoji.demojize(resultado, language='es', delimiters=(" ", " "))

    return resultado
```

###Eliminación de URLs/Emails/Números/HTML


```python
def eliminar_urls_mails_html(texto: str) -> str:
    """
    Elimina URLs, Mails, Html del texto.

    Args:
        texto: Texto a procesar

    Returns:
        Texto sin URLs
    """
    if pd.isna(texto):
        return ""

    # Patrón para detectar URLs, www
    patron = r'https?://\S+|www\.\S+'
    resultado = re.sub(patron, '', str(texto))

    # Patrón para email
    patron_email = r'\b[\w\.-]+@[\w\.-]+\.\w{2,4}\b'
    resultado = re.sub(patron_email, '', str(texto))

    # Eliminar números (incluye decimales)
    resultado = re.sub(r'\d+\.?\d*', '', str(texto))

    # Eliminar etiquetas HTML
    patron_html = r'<.*?>'
    resultado = re.sub(patron_html, '', str(texto))

    return resultado.strip()
```

```python
def corregir_texto(texto: str) -> str:
    """
    Elimina URLs, Mails, Html del texto.

    Args:
        texto: Texto a procesar

    Returns:
        Texto sin URLs
    """
    if pd.isna(texto):
        return ""

    spell = SpellChecker(language='es')  # Español

    palabras = texto.split()
    texto_corregido = []

    for palabra in palabras:
        corregida = spell.correction(palabra)
        texto_corregido.append(corregida if corregida else palabra)

    return ' '.join(texto_corregido)
```

###Eliminación de acentos

```python
def eliminar_acentos(texto: str) -> str:
    """
    Elimina acentos y diacríticos del texto.

    Args:
        texto: Texto con acentos

    Returns:
        Texto sin acentos
    """
    if pd.isna(texto):
        return ""

    # Normalizar el texto y eliminar diacríticos
    texto_nfd = unicodedata.normalize('NFD', str(texto))
    resultado = ''.join(char for char in texto_nfd if unicodedata.category(char) != 'Mn')

    return resultado
```

```python
# ==============================
# 3. Preprocesamiento paso a paso
# ==============================

# 1) Conversión a minúsculas
minus = convertir_minusculas(texto)
print("\n=== Texto en minúsculas ===")
print(minus)
```

```python
# 3) Eliminación de acentos
print("\n=== Texto sin acentos ===")
sin_acentos = eliminar_acentos(texto)
print(sin_acentos)
```

```python
# 4) Tokenización con spaCy
# Eliminar emoticons
sin_emoji = eliminar_emojis(sin_acentos)
print(eliemoji)

doc = nlp(sin_emoji)

tokens = [token.text for token in doc]
print("\n=== Tokens ===")
print(tokens)

# Visualización: Frecuencia de palabras (tokens crudos)
counter = Counter(tokens)
most_common = counter.most_common(10)
palabras, freqs = zip(*most_common)
plt.figure(figsize=(10,5))
plt.bar(palabras, freqs)
plt.title("Frecuencia de palabras (tokens crudos)")
plt.show()

# WordCloud de tokens crudos
text_joined = ' '.join(tokens)
wc = WordCloud(width=800, height=400, background_color="white").generate(text_joined)
plt.figure(figsize=(10,5))
plt.imshow(wc, interpolation="bilinear")
plt.axis("off")
plt.title("WordCloud de tokens crudos")
plt.show()

```

```python
# 5) Lematización con spaCy
lemmas = [token.lemma_ for token in doc if not token.is_punct and not token.is_space]
print("\n=== Lemas ===")
print(lemmas)

# Visualización: Frecuencia de palabras (lematizadas)
counter = Counter(lemmas)
most_common = counter.most_common(10)
palabras, freqs = zip(*most_common)
plt.figure(figsize=(10,5))
plt.bar(palabras, freqs)
plt.xlabel("Palabras", fontsize=12)
plt.ylabel("Frecuencia", fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.title("Frecuencia de palabras (lematizadas)")
plt.show()

# WordCloud de lemas
text_joined = ' '.join(lemmas)
wc = WordCloud(width=800, height=400, background_color="white").generate(text_joined)
plt.figure(figsize=(10,5))
plt.imshow(wc, interpolation="bilinear")
plt.axis("off")
plt.title("WordCloud de lemas")
plt.show()

```

```python
# 6) Eliminación de stopwords
stop_words = set(stopwords.words('spanish'))
tokens_clean = [word for word in lemmas if word not in stop_words]
print("\n=== Tokens limpios (sin stopwords) ===")
print(tokens_clean)

# Visualización: Palabras más importantes (sin stopwords)
counter = Counter(tokens_clean)
most_common = counter.most_common(10)
palabras, freqs = zip(*most_common)
plt.figure(figsize=(12, 6))
plt.bar(palabras, freqs, color='steelblue', alpha=0.8)
plt.title("Palabras más importantes (sin stopwords)", fontsize=14, fontweight='bold')
plt.xlabel("Palabras", fontsize=12)
plt.ylabel("Frecuencia", fontsize=12)
plt.xticks(rotation=45, ha='right')  # Rotar etiquetas 45 grados y alinear a la derecha
plt.tight_layout()
plt.grid(axis='y', alpha=0.3, linestyle='--')
plt.show()

# WordCloud de palabras limpias
text_joined = ' '.join(tokens_clean)
wc = WordCloud(width=800, height=400, background_color="white").generate(text_joined)
plt.figure(figsize=(10,5))
plt.imshow(wc, interpolation="bilinear")
plt.axis("off")
plt.title("WordCloud de palabras limpias")
plt.show()
```
