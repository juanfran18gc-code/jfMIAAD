# Práctica #5 - Unidad 1. Uso de máscaras booleanas
Crea un arreglo unidimensional con 15 números aleatorios entre 1 y 50. Luego:

1.  Selecciona los elementos mayores que 25 utilizando una máscara booleana.
2.  Cuenta cuántos elementos son mayores que 25.

```python
import numpy as np

print("Arreglo de 15 elementos con valores enteros aleatorios entre 1 y 50")
#Agregamos una semilla 2 para generar los mismos numeros aleatorios= reproducibilidad
np.random.seed(2)

#Generamos un arreglo de números aleatorios entre 1 y 50
arreglo = np.random.randint(1,51,size=15)
#Imprimimos el arreglo
print(arreglo)

#Seleccionar los elementos mayores a 25 con mascara
print("Elementos mayores a 25: ", arreglo[arreglo > 25])

#Contar los elementos mayores a 25 con mascara
cuenta = arreglo[arreglo > 25]
print("Cuenta de mayores a 25: ", cuenta.size)
```
