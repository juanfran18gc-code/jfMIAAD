# Práctica 2: Expresiones Regulares

En esta practica, se te pide que diseñes un código en Google Colab para analizar el archivo informe_acciones.txt. El archivo contiene varios correos electrónicos en donde se informa el valor de las acciones de diferentes empresas. El código deberá extraer el valor actual de las acciones que aparecen en cada correo electrónico e imprimirlo a pantalla. En este ejercicio deberás utilizar Expresiones Regulares


```python
# Montar Google Drive
from google.colab import drive
drive.mount('/content/drive')
```

```python
#Importamos biblioteca para expresiones regulares
import re

# Abre el archivo informe_acciones.txt
archivo = open('/content/drive/MyDrive/ClassFiles/informe_acciones.txt')

for linea in archivo:
    #Elimina espacios y salto de línea
    linea = linea.rstrip()
    #Expresion regular busca que la cadena inicie con "Valor actual:", luego espacio en blanco, seguido de signo "$"
    #seguido de cualquier numero, luego punto y uno o mas digitos
    acciones = re.findall('^Valor actual:\s[$]([0-9.]+\d)', linea)
    if len(acciones) > 0:
        #Convetir la lista a valores tipo float
        vacciones =[float(x) for x in acciones]
        #Imprimir valores de las acciones
        print("Valor:", vacciones)
#Cerramos el archivo para evitar perdida de datos
archivo.close()
```
