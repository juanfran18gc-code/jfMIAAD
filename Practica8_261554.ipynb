```python
import seaborn as sns
import pandas as pd
#import numpy as np
import matplotlib.pyplot as plt

#1 Cargar el dataset Iris
dftips= sns.load_dataset('tips')
#Imprimir las primeras y ultimas 5 filas
print(dftips.head())
print(dftips.tail())

#Obtener información general sobre los datos (nombres de columnas, tipos de datos, valores nulos, etc.).
print(dftips.info())

#Descripción y estadisticas de valores numericas
print(f"\nResumen de las columnas numéricas:\n")
print(dftips.describe())

#Verificar la existencia de valores nulos o datos faltantes.
print(f"\nResumen {dftips.isnull().sum()}")
```

```python
#Crear mapa de calor para correlación entre las variables numéricas (total_bill, tip, size).
#Generar Matriz de correlación
columnas = dftips[['total_bill', 'tip', 'size']]
correlacion = columnas.corr()
print("Matriz de correlación:\n", correlacion)

#Crea mapa de calor
sns.set_palette("muted")
sns.set_style("darkgrid")
sns.heatmap(correlacion, cmap="coolwarm", annot=True)

plt.title("Mapa de calor - Correlación entre total_bill, tip y size")
plt.show()
```

```python
#Crear diagrama de violín para comparar la distribución de las propinas en cada uno de los días
sns.violinplot(data=dftips, x="day", y="tip", hue='day', palette="plasma", inner="quartile")

# Colocar etiquetas y título
plt.title("Distribución de las propinas por día")
plt.xlabel("Día")
plt.ylabel("Propina")
plt.show()
```

```python
#Crear un gráfico de dispersión con total_bill en el eje X y tip en el eje Y. Color diferente para cada día

sns.scatterplot(data=dftips, x='total_bill', y='tip', hue='day', style='day', s=50)

# Configurar el gráfico
plt.title('Relación entre Cuenta Total y Propina por Día')
plt.xlabel('Cuenta Total $')
plt.ylabel('Propinas $')
plt.grid(True)
plt.legend(title='Día de la semana')

# Mostrar el gráfico
plt.show()
```

```python
#Crear gráfico de barras para promedio de las propinas por género (hombres y mujeres).
sns.barplot(data=dftips, x='sex', y='tip', hue='sex',palette='magma')

# Añadir etiquetas y título
plt.title('Promedio de Propinas por Género', fontsize=10)
plt.xlabel('Género', fontsize=10)
plt.ylabel('Promedio de Propinas $', fontsize=10)
plt.grid(axis='y', linestyle=':', alpha=0.5)

plt.show()
```
