# Práctica 1: Manejo de Archivos

En esta practica, se te pide que diseñes un código en Google Colab para analizar el archivo mbox.txt (revisa el laboratorio 1). El código deberá extraer las mediciones contenidas en cada línea que contiene la cadena X-DSPAM-Confidence.
Objetivos:
1. Tu código debe extraer únicamente los datos numéricos de las líneas que
contienen la cadena X-DSPAM-Confidence en el archivo mbox.txt.
2. Para extraer los datos númericos usa las intrucciones vistas en el laboratorio 1 (startswith). Aquí no se usan expresiones regulares
3. Los datos numéricos los deberás guardar en una lista y posteriormente mostrar los resultados de cada medición encontrada.
4. Explica brevemente en tu Google Colab qué realiza el código.
5. Calcular y mostrar en pantalla:
*  	La cantidad de datos extraídos.
*		La sumatoria de los datos.
*		El promedio de los datos.
*		La varianza de los datos.

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
# Abre el archivo mbox.txt
archivo = open('/content/drive/MyDrive/ClassFiles/mbox.txt')
#Inicializamos lista vacia y suma
valores = []
suma=0
for linea in archivo:
    linea = linea.rstrip()
    # Ignorar 'líneas que no nos interesan'
    if not linea.startswith('X-DSPAM-Confidence:'):
        continue
    # Va agregando a la lista solo los valores que coinciden con la cadena 'X-DSPAM-Confidence:', split para solo tomar la cadena del valor númerico
    valores.append(linea.split(':')[1])
#Imprime el titulo de la lista
print("Datos de la lista:")
#Iteramos en la lista valores para obtener la suma de los valores de la lista
for numero in valores:
    print(numero)
    #Adiciona a la suma los valores numericos tipo float
    suma=suma+float(numero)
#Calculamos el promedio de los valores numericos de la lista
promedio=suma/len(valores)
#Calculamos la varianza
varianza=sum([(float(numero)-promedio)**2 for numero in valores])/len(valores)

#Imprimimos los resultados solicitados
print("Cantidad de datos extraídos: ",len(valores))
print("La sumatoria de los datos es: ",suma)
print("El promedio de los datos es: ",promedio)
print("La varianza de los datos es: ", varianza)
```
