# 1. Operaciones básicas
**Objetivo: Crear y manipular arrays.**


Instrucciones:
1. Crea un array de 10 elementos con valores entre 1 y 10.
2. Calcula la media, mediana, y desviación estándar de los valores.
3. Multiplica cada valor por 2.

```python
import numpy as np

# 1. Crear un array de 10 elementos con valores entre 1 y 10

np.random.seed(7) # Para reproducibilidad
arr = np.random.randint(1, 11, 10)

print("Array inicial:", arr)

# 2. Calcular estadísticas
media = np.mean(arr)
mediana = np.median(arr)
desviacion_estandar = np.std(arr)

print("Estadísticas:")
print(f"Media: {media}")
print(f"Mediana: {mediana}")
print(f"Desviación estándar: {desviacion_estandar:.4f}")


# 3. Multiplicar cada valor por 2
arr_multiplicado = arr * 2

print("\nArray multiplicado por 2:",arr_multiplicado)
```

# 2. Operaciones con DataFrames
**Objetivo: Crear y manipular un DataFrame.**


Instrucciones:
1. Crea un DataFrame con 3 columnas (A, B, C) y 5 filas con números aleatorios enteros.
2. Renombra las columnas a Col1, Col2, Col3.
3. Añade una nueva columna que sea la suma de las columnas Col1 y Col2.
4. Elimina todas las filas donde el valor de Col3 sea menor que 10.

```python
import pandas as pd
import numpy as np

# 1. Crear DataFrame con 3 columnas (A, B, C) y 5 filas con números aleatorios enteros
np.random.seed(7)  # Para reproducibilidad
df = pd.DataFrame({
    'A': np.random.randint(1, 20, 5),
    'B': np.random.randint(1, 20, 5),
    'C': np.random.randint(1, 20, 5)
})

print("DataFrame original:\n",df)


# 2. Renombrar las columnas a Col1, Col2, Col3
df = df.rename(columns={'A': 'Col1', 'B': 'Col2', 'C': 'Col3'})

print("\nDataFrame con columnas renombradas:\n",df)

# 3. Añadir una nueva columna que sea la suma de Col1 y Col2
df['Suma'] = df['Col1'] + df['Col2']

print("\nDataFrame con columna de suma:\n",df)

# 4. Eliminar todas las filas donde el valor de Col3 sea menor que 10
df_filtrado = df[df['Col3'] >= 10]

print("\nDataFrame después de eliminar filas donde Col3 < 10:\n",df_filtrado)
```

# 4. Análisis de series temporales
**Objetivo: Trabajar con fechas y series temporales.**


Instrucciones:
1. Crea un DataFrame con una columna de fechas (usa pd.date_range() para generar las fechas).
2. Genera 100 valores aleatorios asociados a estas fechas.
3. Calcula la media móvil de los valores con una ventana de 5 días.

```python
import pandas as pd
import numpy as np

# 1. Crear DataFrame con una columna de fechas
fechas = pd.date_range(start='2025-11-01', periods=100, freq='D')

# 2. Generar 100 valores aleatorios asociados a estas fechas
np.random.seed(7)  # Para reproducibilidad
val = np.random.randint(10, 100, 100)

# Crear el DataFrame
df = pd.DataFrame({
    'Fecha': fechas,
    'Valor': val
})

print("DataFrame inicial (primeras 5 filas):\n",df.head(5))

# 3. Calcular la media móvil con una ventana de 5 días
df['Media_Movil_5'] = df['Valor'].rolling(window=5).mean()

print("\nDataFrame con media móvil de 5 días:\n",df)
```

# 6. Gráfico de Barras
**Objetivo: Visualizar datos categóricos.**


Instrucciones:
1. Carga el dataset tips de Seaborn.
2. Realiza un gráfico de barras que muestre el total de las cuenta (total_bill) agrupadas por día.
3. Añade etiquetas a los ejes y un título.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# 1. Cargar el dataset tips de Seaborn
df = sns.load_dataset('tips')

print("Dataset 'tips'(primeras 10 filas):\n", df.head(10))


# Agrupar por día y calcular el total de las cuentas
total_cuentas = df.groupby('day', observed=False)['total_bill'].sum().reset_index()

print("\nTotal de cuentas agrupadas por día:\n",total_cuentas)


# 2. Crear el gráfico de barras
plt.figure(figsize=(10, 6))

sns.barplot(data=total_cuentas, x='day', y='total_bill',
            palette='magma', hue='day', legend=False)

# 3. Añadir etiquetas a los ejes y un título
plt.title('Total de Cuentas Agrupadas por Día',
          fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Día de la Semana', fontsize=12)
plt.ylabel('Total de Cuentas ($)', fontsize=12)
plt.tight_layout()
plt.show()
```

# 7. Histograma
**Objetivo: Visualizar la distribución de datos numéricos.**

Instrucciones:
1. Genera 1000 números aleatorios con distribución normal (media = 0, desviación estándar = 1).
2. Crea un histograma para visualizar la distribución.
3. Ajusta el número de bins del histograma para que la visualización sea clara.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
# 1. Generar 1000 números aleatorios con distribución normal
np.random.seed(15)  # Para reproducibilidad
media = 0
desviacion_std = 1
n_datos = 1000

datos = np.random.normal(loc=media, scale=desviacion_std, size=n_datos)

#print("Datos:\n",datos)

# 2 y 3. Crear histograma con número ajustado de bins
plt.figure(figsize=(12, 7))

sns.histplot(datos, bins=30, alpha=0.7, color='steelblue',
        edgecolor='black', linewidth=0.5,
        label='Datos', kde=True)

# Añadir título y etiquetas
plt.title('Histograma de Distribución Normal\n(1000 valores aleatorios)',
          fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Valor', fontsize=12)
plt.ylabel('Densidad', fontsize=12)

# Añadir leyenda
plt.legend(loc='upper right', fontsize=10, framealpha=0.9)
plt.tight_layout()
plt.show()
```

# 9. Heatmap
**Objetivo: Visualizar la correlación entre variables numéricas.**

Instrucciones:
1. Usa el dataset iris de Seaborn.
2. Calcula la matriz de correlación de las variables.
3. Crea un heatmap para visualizar la matriz de correlación.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# 1. Cargar el dataset iris de Seaborn
df = sns.load_dataset('iris')

print("Dataset 'iris' (primeras 5 filas):\n", df.head())


# 2. Calcular la matriz de correlación de las variables numéricas
# Seleccionamos solo las columnas numéricas
correlacion = df[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']].corr()

print("Matriz de Correlación:\n", correlacion)


# 3. Crear un heatmap para visualizar la matriz de correlación
plt.figure(figsize=(10, 8))

sns.heatmap(correlacion,
            annot=True,
            fmt='.2f',
            cmap='coolwarm',
            center=0,
            square=True,
            linewidths=1,
            cbar_kws={'shrink': 0.8},
            vmin=-1, vmax=1)

# Añadir título
plt.title('Matriz de Correlación - Dataset Iris',
          fontsize=16, fontweight='bold', pad=20)

plt.tight_layout()
plt.show()
```

# 10. Gráfico de Línea con Media Móvil
**Objetivo: Visualizar series temporales.**

Instrucciones:
1. Genera una serie temporal de 100 días con valores aleatorios.
2. Calcula la media móvil de los últimos 10 días y gráficala junto a la serie original.
3. Usa diferentes colores para la serie original y la media móvil.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Generar una serie temporal de 100 días con valores aleatorios
fechas = pd.date_range(start='2025-11-01', periods=100, freq='D')

# 2. Generar 100 valores aleatorios asociados a estas fechas
np.random.seed(15)  # Para reproducibilidad
val = np.random.randint(10, 100, 100)

# Crear el DataFrame
df = pd.DataFrame({
    'Fecha': fechas,
    'Valor': val
})

print("DataFrame inicial (primeras 15 filas):\n",df.head(15))

# 2. Calcular la media móvil de los últimos 10 días
df['Media_Movil_10'] = df['Valor'].rolling(window=10).mean()

print("Media móvil calculada (ventana de 10 días)")


# 3. Graficar la serie original y la media móvil con diferentes colores
plt.figure(figsize=(14, 7))

plt.plot(df['Fecha'], df['Valor'],
         color='steelblue',
         linewidth=1.5,
         alpha=0.7,
         marker='o',
         markersize=3,
         label='Serie Original')


plt.plot(df['Fecha'], df['Media_Movil_10'],
         color='red',
         linewidth=2.5,
         label='Media Móvil (10 días)')

# Añadir título y etiquetas
plt.title('Serie Temporal con Media Móvil de 10 Días',
          fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Fecha', fontsize=12)
plt.ylabel('Valor', fontsize=12)

# Añadir leyenda
plt.legend(loc='upper right', fontsize=11, framealpha=0.9)

# Rotar las fechas para mejor visualización
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

# 11. Datos Temporales con Análisis Estadístico
**Objetivo: Trabajar con series temporales y calcular estadísticas.**

Instrucciones:
1. Genera una serie temporal de 200 días con valores aleatorios.
2. Calcula la media y la desviación estándar de la serie temporal.
3. Visualiza la serie junto con las bandas de desviación estándar (es decir, la media ± 1 desviación estándar).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Generar una serie temporal de 200 días con valores aleatorios
fechas = pd.date_range(start='2025-11-01', periods=200, freq='D')

# 2. Generar 100 valores aleatorios asociados a estas fechas
np.random.seed(15)  # Para reproducibilidad
val = np.random.randint(10, 100, 200)

# Crear el DataFrame
df = pd.DataFrame({
    'Fecha': fechas,
    'Valor': val
})

print("DataFrame inicial (primeras 15 filas):\n",df.head(15))

# 2. Calcular la media y la desviación estándar de la serie temporal
media = df['Valor'].mean()
desviacion_std = df['Valor'].std()

print("Estadísticas de la serie temporal:\n")
print(f"Media: {media:.2f}")
print(f"Desviación estándar: {desviacion_std:.2f}")

# Calcular las bandas de desviación estándar
banda_superior = media + desviacion_std
banda_inferior = media - desviacion_std

# Contar valores dentro y fuera de las bandas
dentro_bandas = df[(df['Valor'] >= banda_inferior) & (df['Valor'] <= banda_superior)]
porcentaje_dentro = (len(dentro_bandas) / len(df)) * 100

# 3. Visualizar la serie con las bandas de desviación estándar
plt.figure(figsize=(15, 8))

# Graficar la serie original
plt.plot(df['Fecha'], df['Valor'],
         color='steelblue',
         linewidth=1.5,
         alpha=0.8,
         label='Serie Original')

# Graficar la línea de media
plt.axhline(y=media,
            color='darkgreen',
            linestyle='--',
            linewidth=2,
            label=f'Media: {media:.2f}')

# Graficar las bandas de desviación estándar
plt.axhline(y=banda_superior,
            color='red',
            linestyle=':',
            linewidth=2,
            label=f'μ + σ: {banda_superior:.2f}')

plt.axhline(y=banda_inferior,
            color='red',
            linestyle=':',
            linewidth=2,
            label=f'μ - σ: {banda_inferior:.2f}')

# Añadir título y etiquetas
plt.title('Serie Temporal con Bandas de Desviación Estándar (μ ± σ)',
          fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Fecha', fontsize=12)
plt.ylabel('Valor', fontsize=12)

# Añadir leyenda
plt.legend(loc='upper right', fontsize=10, framealpha=0.95)

# Rotar las fechas
plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
```
