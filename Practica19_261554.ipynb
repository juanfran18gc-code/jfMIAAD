# **Ejercicio 01:** Duplicidad Parcial Una columna: Contar y Filtrar los registros duplicados en Duration

1.  Carga el conjunto de datos dirty.csv y verifica que su carga sea correcta.
2.  Cuenta la cantidad de registros duplicados utilizando solo la información de columna Duration e imprime
3.  Utiliza value_counts() e imprime la cantidad de registros que hay por cada valor único.
4.  Filtra los datos y obten los registros sin duplicados

```python
import pandas as pd
#Cargar el dataset dirty.csv
df = pd.read_csv('/content/drive/MyDrive/Unidad02/dirtydata.csv')

#Observar las columnas y datos del dataframe así como los tipos de datos
print(df.head(5))
print(df.dtypes)
```

```python
#Contar registros duplicados
contar_duplicados_duration = df.duplicated(subset=['Duration']).sum()
print(f'Cantidad de Registros duplicados:',contar_duplicados_duration)
```

```python
#Mostrar frecuencia de valores en columna
print(f'Frecuencia de valores únicos:\n',df['Duration'].value_counts())
```

```python
#Obtener registros sin duplicados
duration_sin_duplicados = df[~df.duplicated(subset='Duration', keep=False)]
print(f'Registros sin duplicados:\n',duration_sin_duplicados)
```

# **Ejercicio 02:** Duplicidad Parcial Dos columnaa: Contar y Filtrar los registros duplicados en Pulse y MaxPulse

1.  Carga el conjunto de datos dirty.csv y verifica que su carga sea correcta.
2.  Cuenta la cantidad de registros duplicados utilizando la información de las columnas Pulse y Maxpulse
3.  Utiliza value_counts() e imprime la cantidad de registros que hay por cada valor único.
4.  Filtra los datos y obten los registros cuando keep = 'last'

```python
import pandas as pd
#Cargar el dataset dirty.csv
df = pd.read_csv('/content/drive/MyDrive/Unidad02/dirtydata.csv')

#Observar las columnas y datos del dataframe así como los tipos de datos
print(df.head(5))
print(df.dtypes)
```

```python
#Contar registros duplicados de dos columnas
contar_duplicados_dos_col = df.duplicated(subset=['Pulse','Maxpulse']).sum()
print(f'Cantidad de Registros duplicados en dos columnas:',contar_duplicados_dos_col)
```

# **Ejercicio 03:** Duplicidad Total o Exacta: Contar y Filtrar los registros duplicados

1.  Carga el conjunto de datos dirty.csv y verifica que su carga sea correcta.
2.  Cuenta la cantidad de registros duplicados exactos (todas las columnas)
3.  Utiliza sum() e imprime la cantidad de registros con duplicidad exacta
4.  Filtra los datos y obten los registros cuando keep = 'last'

```python
import pandas as pd
#Cargar el dataset dirty.csv
df = pd.read_csv('/content/drive/MyDrive/Unidad02/dirtydata.csv')

#Observar las columnas y datos del dataframe así como los tipos de datos
print(df.head(5))
print(df.dtypes)
```

# **Ejercicio 04:** Cálculo de la media con duplicados y sin duplicados

1.  Carga el conjunto de datos dirty.csv y verifica que su carga sea correcta.
2.  Calcula e imprime la media de las columnas Pulse y MaxPulse
3.  Identifica los valores duplicados en las columnas Pulse y MaxPulse (Duplicidad Parcial) y elimina los registro repetidos. Quédate con las últimas ocurrencias
4.  Utiliza el conjunto anterior sin datos duplicados (recuerda que solo quedaron las últimas ocurrencias de los datos) y calcula la media de las columnas Pulse y MaxPulse

```python
import pandas as pd
#Cargar el dataset dirty.csv
df = pd.read_csv('/content/drive/MyDrive/Unidad02/dirtydata.csv')

#Observar las columnas y datos del dataframe así como los tipos de datos
print(df.head(5))
print(df.dtypes)
```

```python
#Calcular media de las columnas
print(f'Promedio de Pulse:\n',df['Pulse'].mean())
print(f'Promedio de Maxpulse:\n',df['Maxpulse'].mean())
```

```python
#Eliminar duplicados de Pulse y Maxpulse
df_sin_duplicados = df.drop_duplicates(subset=['Pulse','Maxpulse'], keep='last')
print(f'Registros sin duplicados con last:\n',df_sin_duplicados)
```

```python
#Calcular media de las columnas sin duplicados
print(f'Promedio de Pulse:\n',df_sin_duplicados['Pulse'].mean())
print(f'Promedio de Maxpulse:\n',df_sin_duplicados['Maxpulse'].mean())
```

##Conclusiones
* Cargamos el dataset sin convertir a datatime la columna Date
* Para detectar datos duplicados usamos `df.duplicated(subset=None,keep='first')` 'first' para la primera ocurrencia o 'last' para la última
* Para duplicados parciales usar `df.duplicated(subset=['Columna'],keep=False)` False para todas las ocurrencias
* Contar duplicados con `df.duplicated().sum()`
* Contar duplicados parciales con `df.duplicated(subset=['Columna'],keep=False).sum()`
* Frecuencia de los valores de una columna usar `df['Columna'].value_counts()`
* Filtrar duplicados `df[df.duplicated(subset=None, keep='last')]`
