# Práctica #11: Seaborn y Subplots

**Objetivo**

Utilizar Seaborn junto con plt.subplots() para visualizar diferentes aspectos de un conjunto de datos en una sola figura.

Instrucciones:
1.  * Carga el dataset "penguins" de Seaborn. Utiliza df = sns.load_dataset("penguins")
    * Usa plt.subplots() para crear una figura con 2 filas y 2 columnas.
    * Genera los siguientes gráficos:
      * (0,0): sns.histplot(): Histograma de las longitudes de aletas.
      * (0,1): sns.boxplot(): Boxplot comparando la masa corporal por especie.
      * (1,0): sns.scatterplot(): Dispersión entre masa corporal y largo de aletas.
      * (1,1): sns.violinplot(): Violin plot de la masa corporal según la especie.
    * Personaliza los gráficos con títulos y colores adecuados.
    * Ajusta el diseño con plt.tight_layout().

```python
import matplotlib.pyplot as plt
import seaborn as sns
#Cargar el data set penguins
df = sns.load_dataset("penguins")
#print(df.head())

# Crear la figura y los subplots 2 X 2
fig, ax = plt.subplots(2, 2, figsize=(12, 8))

#  Histograma de las longitudes de aletas.
sns.histplot(df['flipper_length_mm'], bins=30, kde=True,ax=ax[0, 0])
ax[0, 0].set_title("Longitudes de aletas", color='blue')
ax[0, 0].set_xlabel('Longitud de la aleta')
ax[0, 0].set_ylabel('Frecuencia')

#  Boxplot comparando la masa corporal por especie
sns.boxplot(data=df, x='species', y='body_mass_g',notch=True, hue='species', palette="coolwarm", ax=ax[0,1])
ax[0, 1].set_title("Masa corporal por especie", color='blue')
ax[0, 1].set_xlabel('Especies')
ax[0, 1].set_ylabel('Masa corporal')

#  Dispersión entre masa corporal y largo de aletas
sns.scatterplot(data=df, x='flipper_length_mm', y='body_mass_g', size=10, marker='s', ax=ax[1, 0])
ax[1, 0].set_title("Relación Masa corporal vs Largo de aletas", color='blue')
ax[1, 0].set_xlabel('Longitud de la aleta')
ax[1, 0].set_ylabel('Masa corporal')

#  Violin plot de la masa corporal según la especie
sns.violinplot(data=df, x='species', y='body_mass_g', hue='species', palette='muted', ax=ax[1, 1])
ax[1, 1].set_title("Masa corporal por especie", color='blue')
ax[1, 1].set_xlabel('Especies')
ax[1, 1].set_ylabel('Masa corporal')

plt.tight_layout()
plt.show()
```
