```python
import pandas as pd
# Leer el archivo CSV
# La separación por default es coma (,)
df = pd.read_csv('/content/drive/MyDrive/ClassFiles/Titanic-Dataset.csv')

# Ejercicio 1 Análisis de la distribución de supervivencia por combinación de sexo y clase del pasajero.
sup = df.groupby(['Sex','Pclass'])['Survived'].mean()
tasaalta = sup.idxmax()
tasabaja = sup.idxmin()
#Calcular la proporción de supervivencia para cada combinación de 'Sex' y 'Pclass'.
print("Supervivencia 'Sex' y 'Pclass': \n", sup )
#Identificar tasa de supervivencia más alta.
print(f"Tasa de supervivencia más alta': \n {tasaalta[0], int(tasaalta[1])} tasa: {sup.max()} ")
#Identificar tasa de supervivencia más baja.
print(f"Tasa de supervivencia más baja': \n {tasabaja[0], int(tasabaja[1])} tasa: {sup.min()} ")
```

```python
#Ejercicio 2: Identificación de familias grandes a bordo.
#Crear nueva columna FamilySize
df['FamilySize'] = df['SibSp'] + df['Parch']
print(df['FamilySize'])

#Identificar "familia grande" donde 'FamilySize' es mayor a 3.
bigfam= df[df['FamilySize'] > 3]
print(f"Familia grande: {bigfam}")

#Calcular el número de pasajeros en familias grandes.
print(f"Número de pasajeros: {len(bigfam)}")

#Calcular la proporción de supervivencia entre los pasajeros que pertenecen a familias grandes.
print(f"Supervivencia familias grandes: {bigfam['Survived'].mean()}")
```

```python
#Ejercicio 3: Segmentación por grupos de edad.
#Clasificar pasajeros en las siguientes categorías de edad
#Menor de edad (< 18) y mayor de edad (>=18)

#Funcion para clasificar edades
def clasificar(edad):
  if edad < 18:
    return "Menor de edad"
  elif edad >= 18:
    return "Mayor de edad"
  else:
    return "Edad desconocida"

df['ClasificarEdad'] = df['Age'].apply(clasificar)
print(f"Clasificacion de edades:\n {df[['Age','ClasificarEdad']]}")

```

```python
#Ejercicio 4: Comparación entre promedios calculados manualmente y con Pandas
#Calcular el promedio de las columnas 'Age' y 'Fare', ignorando valores nulos.
import numpy as np

#Carlcular promedios de Edad y Tarifa con numpy
coledad = df['Age']
coltarifa = df['Fare']
promEdadnp = coledad.dropna().to_numpy()
promTarifanp = coltarifa.dropna().to_numpy()
print(f"Promedio Edad con numpy:\n {promEdadnp.mean()}")
print(f"Promedio Tarifa con numpy:\n {promTarifanp.mean()}")

#Carlcular promedios de Edad y Tarifa con pandas
promEdad = df['Age'].mean()
promTarifa = df['Fare'].mean()

print(f"Promedio Edad:\n {promEdad}")
print(f"Promedio Tarifa:\n {promTarifa}")
#Compara estos valores con los promedios obtenidos utilizando los métodos nativos de Pandas.
#Verifica que los resultados sean consistentes.
if float(promEdadnp.mean()) == float(promEdad.mean()) and float(promTarifanp.mean()) == float(promTarifa.mean()):
  print("Resultados consistentes")
else:
  print("Resultados no consistentes")
```

```python
#Ejercicio 5. Creación de intervalos de clase usando NumPy y análisis con Pandas
#Divide la columna 'Fare' en 5 intervalos equidistantes utilizando la función numpy.linspace

#Obtener mínimo y máximo de tarifa
tarifamin = df['Fare'].min()
tarifamax = df['Fare'].max()
#Calcular intervalos
intervalo = np.linspace(tarifamin, tarifamax, 6)
print(f"Los intervalos son: {intervalo}")

#Crea una nueva columna en el DataFrame que asigne a cada pasajero el intervalo correspondiente de su tarifa.
df['IntervaloTarifa'] = pd.cut(df['Fare'], bins=intervalo, include_lowest=True)
print(df.head())
print(df.tail())

#Calcula el número de pasajeros en cada intervalo utilizando Pandas y la proporción de supervivientes por intervalo.
#Agrupar por IntervaloTarifa sin considerar los nan, observed False para mostrar un intervalo que en el que no existen pasajeros
sup = df[df['IntervaloTarifa'].notna()].groupby('IntervaloTarifa', observed=False).agg(
    cantidadpasajeros=('IntervaloTarifa', 'size'),
    proporcionsupervivientes=('Survived', 'mean')
)

print(f"Supervivientes: \n", sup)
```
